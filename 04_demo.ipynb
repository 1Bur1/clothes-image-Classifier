# 04 Demo - Live Clothing Classifier
Upload a clothing photo and the model predicts the category.
**Run cells top to bottom.**

Prerequisite: run `01_train.ipynb` first so `models/fashion_cnn.h5` exists.

## Step 1 - Install Dependencies

```python
import subprocess, sys
# sys.executable ensures pip works on Windows
subprocess.run([sys.executable, '-m', 'pip', 'install', 'gradio', 'pillow', '-q'], check=True)
```

## Step 2 - Load Model

```python
import numpy as np
import gradio as gr
from tensorflow import keras
from PIL import Image

CLASS_NAMES = [
    'T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
    'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot'
]

model = keras.models.load_model('/content/fashion_cnn.h5')
print(f'Model loaded. Input shape: {model.input_shape}')
```

## Step 3 - Prediction Function

Preprocessing: grayscale → resize 28x28 → normalize → reshape(1,28,28,1)

```python
def predict(image):
    if image is None:
        return {n: 0.0 for n in CLASS_NAMES}
    img = Image.fromarray(image).convert('L').resize((28, 28))
    arr = np.array(img) / 255.0
    arr = arr.reshape(1, 28, 28, 1)
    probs = model.predict(arr, verbose=0)[0]
    return {CLASS_NAMES[i]: float(probs[i]) for i in range(10)}

print('predict() ready.')
```

## Step 4 - Launch Web App

Starts a local server and opens the browser at `http://127.0.0.1:7860`.

```python
gr.Interface(
    fn=predict,
    inputs=gr.Image(type='numpy', label='Upload a clothing photo'),
    outputs=gr.Label(num_top_classes=3, label='Top 3 Predictions'),
    title='Fashion-MNIST Clothing Classifier',
    description='Categories: ' + ', '.join(CLASS_NAMES),
).launch()
```
